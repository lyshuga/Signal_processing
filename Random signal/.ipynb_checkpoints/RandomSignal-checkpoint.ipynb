# 1. Noise analysis
In this section we will analyse the noise files given for this exercise

## Noise 1 

### Periodogram

```python
samplerate, data = wavfile.read("noise1.wav") #Read the file

f, Pxx_den = periodogram(data, samplerate)
plt.semilogy(f, Pxx_den)
plt.ylim([1e-7, 1e2])
plt.xlabel('frequency [Hz]')
plt.ylabel('PSD [V**2/Hz]')
plt.show()
```
